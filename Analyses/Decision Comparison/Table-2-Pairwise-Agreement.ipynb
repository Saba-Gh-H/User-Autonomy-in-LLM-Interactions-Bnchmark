# Table 2: Pairwise model agreement

This notebook reproduces all 66 pairwise exact-agreement and unweighted Cohen's kappa values from the original stored T=0 baseline responses. Responses are paired by `scenario_code`; the five highest and five lowest kappa values form manuscript Table 2.

The accepted Qwen2.5 72B CSV contains six copied scenario-code metadata values in its final six rows. The canonical script verifies that exact known pattern and corrects only an in-memory analysis copy. It never changes the stored response CSV or any response text. Aya Expanse 32B has 119 responses because `LAW01_EPI02` is missing.

```python
import importlib.util
from pathlib import Path

start = Path.cwd().resolve()
repo_root = next(
    candidate
    for candidate in (start, *start.parents)
    if (candidate / 'LLM Normal Test').is_dir()
    and (candidate / 'Dataset' / 'Autonomy_Dataset.json').is_file()
)
script_path = repo_root / 'Analyses' / 'Decision Comparison' / 'table2_pairwise_agreement.py'
spec = importlib.util.spec_from_file_location('table2_pairwise_agreement', script_path)
table2_analysis = importlib.util.module_from_spec(spec)
spec.loader.exec_module(table2_analysis)

results = table2_analysis.run_analysis(repo_root=repo_root, write_outputs=True)
```

```python
# Input validation: original row counts, unique scenario codes, and parse status.
results['validation']
```

```python
# Complete reproducible manuscript Table 2, displayed at manuscript precision.
table2_display = results['table2'][
    ['rank_group', 'rank', 'model_pair', 'n_compared', 'agreement', 'cohen_kappa']
].copy()
table2_display.style.format({'agreement': '{:.3f}', 'cohen_kappa': '{:.3f}'})
```
