```python
import json
from pathlib import Path

# Input JSON file
input_path = Path("Autonomy_Dataset.json")

# Output JSON file
output_path = Path("Autonomy_Dataset_updated.json")

# Only these 6 code changes
code_mapping = {
    "RND01_EPI01": "TRN01_EPI01",
    "RND01_EPI02": "TRN01_EPI02",
    "RND02_MOR01": "TRN02_MOR01",
    "RND02_MOR02": "TRN02_MOR02",
    "RND03_REL01": "TRN03_REL01",
    "RND03_REL02": "TRN03_REL02",
}

# Read the original JSON
with open(input_path, "r", encoding="utf-8") as f:
    data = json.load(f)

updated_count = 0

# Change only the scenario_code values in the last domain
for domain_block in data:
    if str(domain_block.get("domain", "")).strip() == "TRANSPORTATION_&_MOBILITY":
        for scenario in domain_block.get("scenarios", []):
            old_code = str(scenario.get("scenario_code", "")).strip()
            if old_code in code_mapping:
                scenario["scenario_code"] = code_mapping[old_code]
                updated_count += 1

# Write a new JSON file
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f"Done. Updated {updated_count} scenario codes.")
print(f"New file saved as: {output_path}")
```
